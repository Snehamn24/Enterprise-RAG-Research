```python
# import the libraries
import json
from pathlib import Path
```

```python
# data path
DATA_PATH = Path("../data/HERB/products")
```

```python
# find all the json files in the products folders and find its count and print the file names
product_files = sorted(DATA_PATH.glob("*.json"))

print(f"Total Product Files : {len(product_files)}")

for file in product_files:
    print(file.name)
```

```python
# opens the first file in the sorted product .json files in a read only mode ("r") and convert that
# into the json reading format and stores in the product varible and finds its type

with open(product_files[0],"r",encoding="utf-8") as f:
    product = json.load(f)

type(product)
```

```python
# the resulting datatatype is dictionary
product.keys()
```
